# Import Libraries

```python
import pandas as pd
import os
import glob
```

# 1. Define the location of the CSV files

```python
DATA_FOLDER = "../Fifa worldcup data/row"

csv_files = glob.glob(os.path.join(DATA_FOLDER, "*.csv"))

print("=" * 70)
print("FIFA WORLD CUP DATA ANALYSIS")
print("=" * 70)

print(f"\nNumber of CSV files found: {len(csv_files)}")
```

# 2. Check if CSV files were found

```python
if len(csv_files) == 0:
    print("\nNo CSV files found!")
    print("Please check that your CSV files are inside:")
    print(DATA_FOLDER)
    raise FileNotFoundError(f"No CSV files found in {DATA_FOLDER}")
```

# 3. Analyze each CSV file


```python
for file in sorted(csv_files):

    file_name = os.path.basename(file)

    print("\n")
    print("=" * 70)
    print(f"FILE: {file_name}")
    print("=" * 70)

    # Read the CSV file
    df = pd.read_csv(file)

    # --------------------------------------------------------
    # Basic information
    # --------------------------------------------------------

    print("\n--- BASIC INFORMATION ---")

    print("Number of rows:", df.shape[0])
    print("Number of columns:", df.shape[1])

    print("\nColumns:")

    for column in df.columns:
        print(" -", column)

    # --------------------------------------------------------
    # Data types
    # --------------------------------------------------------

    print("\n--- DATA TYPES ---")

    print(df.dtypes)

    # --------------------------------------------------------
    # Missing values
    # --------------------------------------------------------

    print("\n--- MISSING VALUES ---")

    missing = df.isnull().sum()

    # Show only columns that contain missing values
    missing = missing[missing > 0]

    if len(missing) == 0:
        print("No missing values.")

    else:
        print(missing)

    # --------------------------------------------------------
    # Duplicate rows
    # --------------------------------------------------------

    print("\n--- DUPLICATES ---")

    duplicates = df.duplicated().sum()

    print("Duplicate rows:", duplicates)

    # --------------------------------------------------------
    # Number of unique values
    # --------------------------------------------------------

    print("\n--- UNIQUE VALUES ---")

    for column in df.columns:

        unique_count = df[column].nunique(dropna=True)

        print(f"{column}: {unique_count} unique values")

    # --------------------------------------------------------
    # Display the first 5 rows
    # --------------------------------------------------------

    print("\n--- FIRST 5 ROWS ---")

    print(df.head())

    # --------------------------------------------------------
    # Statistical summary
    # --------------------------------------------------------

    print("\n--- STATISTICAL SUMMARY ---")

    print(df.describe(include="all").transpose())

    print("\n")
```
